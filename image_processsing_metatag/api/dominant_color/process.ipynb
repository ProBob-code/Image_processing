```python
import pandas as pd
import requests
import json
from tqdm.auto import tqdm # Import tqdm for the progress bar

# Load the CSV file
try:
    df = pd.read_csv('go_jd_20.csv')
    print("CSV file 'go_jd_20.csv' loaded successfully.")
except FileNotFoundError:
    print("Error: 'go_jd_20.csv' not found. Please ensure the file is in the correct directory.")
    exit()

# API endpoint details
url = "http://192.168.131.170/cp/api/v1/extract-color"
headers = {
    'Content-Type': 'application/x-www-form-urlencoded'
}

# List to store the extracted color information (hex and percentage)
colors_extracted = []

print("\n--- Starting full CSV processing to extract color information ---")

# Iterate through all rows of the DataFrame with a tqdm progress bar
# The 'desc' parameter sets the description for the progress bar
for index, row in tqdm(df.iterrows(), total=df.shape[0], desc="Extracting Colors"):
    thumb_url = row['thumb_url']

    payload = {'product_url': thumb_url}
    files = []

    try:
        # Added a timeout to the request to prevent indefinite waiting
        response = requests.request("POST", url, headers=headers, data=payload, files=files, timeout=10)
        # Raise an HTTPError for bad responses (4xx or 5xx status codes)
        response.raise_for_status() 
        
        try:
            response_json = response.json()
            
            # Navigate to the correct nested structure for the color list
            # The color data is under data -> response -> color
            if 'data' in response_json and 'response' in response_json['data'] and \
               'color' in response_json['data']['response'] and \
               len(response_json['data']['response']['color']) > 0:
                
                first_color = response_json['data']['response']['color'][0]
                first_color_hex = first_color.get('hex')
                first_color_percentage = first_color.get('percentage')

                # Store the combined hex and percentage string
                if first_color_hex and first_color_percentage is not None:
                    colors_extracted.append(f"{first_color_hex} ({first_color_percentage}%)")
                else:
                    # Append None if hex or percentage is missing from the first color object
                    colors_extracted.append(None) 
            else:
                # Append None if color data is not found or is empty within 'data.response'
                colors_extracted.append(None) 
        except json.JSONDecodeError:
            # Print a warning for invalid JSON responses and append None
            print(f"Warning: Response for {thumb_url} was not valid JSON. Appending None.")
            colors_extracted.append(None) 

    except requests.exceptions.ConnectionError as e:
        # Handle connection errors, print details, and append None
        print(f"Connection Error for {thumb_url}: Could not connect to the server. Appending None.")
        print(f"Error details: {e}")
        colors_extracted.append(None)
    except requests.exceptions.Timeout:
        # Handle timeout errors and append None
        print(f"Timeout Error for {thumb_url}: The request timed out after 10 seconds. Appending None.")
        colors_extracted.append(None)
    except requests.exceptions.HTTPError as e:
        # Handle HTTP errors (e.g., 404, 500) and append None
        print(f"HTTP Error for {thumb_url}: {e.response.status_code} - {e.response.text}. Appending None.")
        colors_extracted.append(None)
    except requests.exceptions.RequestException as e:
        # Catch any other requests-related exceptions and append None
        print(f"An unexpected Request Error occurred for {thumb_url}: {e}. Appending None.")
        colors_extracted.append(None)
    except Exception as e:
        # Catch any other unexpected errors during processing and append None
        print(f"An unexpected error occurred during processing {thumb_url}: {e}. Appending None.")
        colors_extracted.append(None)

# Add the new 'color_field' column to the DataFrame
df['color_field'] = colors_extracted

# Define the output filename for the updated CSV
output_filename = 'go_jd_20_with_colors.csv'
# Save the updated DataFrame to a new CSV file without the index
df.to_csv(output_filename, index=False)

print(f"\nProcessing complete. The updated CSV file '{output_filename}' has been saved.")
print("Check the 'color_field' column for the extracted hex and percentage values.")
```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
from torchvision import transforms
from torch.autograd import Variable
import requests
import pandas as pd
import json
from colormap import rgb2hex
import extcolors
import re
import io

# --- New: Function to download U2NET model files ---
def download_u2net_model_files():
    """
    Downloads the U2NET model.py file and provides instructions for u2netp.pth.
    """
    # UPDATED URL for model.py as the previous one was giving a 404 error.
    MODEL_PY_URL = "https://raw.githubusercontent.com/xuebinqin/U-2-Net/master/u2net_test/model.py"
    MODEL_PY_DIR = "model" # New directory for model.py
    MODEL_PY_PATH = os.path.join(MODEL_PY_DIR, "model.py")

    U2NETP_PTH_DIR = "saved_models/u2net"
    U2NETP_PTH_PATH = os.path.join(U2NETP_PTH_DIR, "u2netp.pth")
    
    print("\n--- Checking for U2NET model dependencies ---")

    # Download model.py
    os.makedirs(MODEL_PY_DIR, exist_ok=True)
    if not os.path.exists(MODEL_PY_PATH):
        try:
            print(f"📥 Downloading model.py from: {MODEL_PY_URL}")
            response = requests.get(MODEL_PY_URL, stream=True)
            response.raise_for_status()
            with open(MODEL_PY_PATH, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    f.write(chunk)
            print(f"✅ model.py downloaded to: {MODEL_PY_PATH}")
        except requests.exceptions.RequestException as e:
            print(f"❌ Error downloading model.py: {e}")
            print(f"Please download model.py manually from {MODEL_PY_URL} and place it in the '{MODEL_PY_DIR}' directory.")
    else:
        print(f"✅ model.py already exists at: {MODEL_PY_PATH}")

    # Check for u2netp.pth (often requires manual download)
    os.makedirs(U2NETP_PTH_DIR, exist_ok=True)
    if not os.path.exists(U2NETP_PTH_PATH):
        print(f"\n⚠️ Important: U2NETP pre-trained weights ({U2NETP_PTH_PATH}) not found.")
        print("This file is typically hosted on Google Drive or available in GitHub releases.")
        print("Please download 'u2netp.pth' from the official U2NET GitHub repository releases page:")
        print("  Find releases at: https://github.com/xuebinqin/U-2-Net/releases")
        print(f"  Then place the downloaded file inside the directory: {U2NETP_PTH_DIR}/")
        return False # Indicate that not all files are ready
    else:
        print(f"✅ u2netp.pth already exists at: {U2NETP_PTH_PATH}")
        return True # Indicate all files are ready


# IMPORTANT:
# The U2NETP model requires two external files:
# 1. model.py: Contains the U2NETP model architecture (class U2NETP).
# 2. saved_models/u2net/u2netp.pth: The pre-trained weights for the model.
# These files are NOT part of standard Python libraries and must be placed
# in your working directory as per the original script's structure.

# After `download_u2net_model_files()` is called, we expect 'model.py' to be available.
# We will dynamically add the 'model' directory to Python's path to import U2NETP.
import sys
if "model" not in sys.path:
    sys.path.append("model")

try:
    from model import U2NETP
except ImportError:
    # This block should ideally not be hit if download_u2net_model_files() succeeds,
    # but serves as a final safeguard if path manipulation or download fails.
    print("❌ Critical Error: Could not import U2NETP from model.py.")
    print("Please ensure model.py is correctly downloaded to the 'model' directory.")
    class U2NETP(torch.nn.Module): # Fallback mock model to prevent immediate crash
        def __init__(self, in_ch=3, out_ch=1):
            super().__init__()
            print("U2NETP: Mock model loaded due to import failure. Background removal will not occur.")
        def forward(self, x):
            return torch.zeros_like(x), torch.zeros_like(x), torch.zeros_like(x), \
                   torch.zeros_like(x), torch.zeros_like(x), torch.zeros_like(x), \
                   torch.zeros_like(x)

# --- Part 1: Download Image Function ---
def download_image(url: str, output_dir: str = "downloads", filename: str = None) -> str:
    """
    Downloads an image from a given URL and saves it to a specified directory.

    Args:
        url (str): The URL of the image to download.
        output_dir (str): The directory where the image will be saved. Defaults to "downloads".
        filename (str, optional): The name to save the file as. If None, derives from URL.

    Returns:
        str: The full path to the downloaded image, or None if download fails.
    """
    os.makedirs(output_dir, exist_ok=True)
    if filename is None:
        filename = url.split('/')[-1].split('?')[0] # Basic filename extraction
        if not filename: # Fallback if URL doesn't provide a good filename
            filename = "downloaded_image.png"
    
    file_path = os.path.join(output_dir, filename)

    try:
        print(f"📥 Downloading image from: {url}")
        response = requests.get(url, stream=True)
        response.raise_for_status() # Raise an exception for bad status codes

        with open(file_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"✅ Image downloaded to: {file_path}")
        return file_path
    except requests.exceptions.RequestException as e:
        print(f"❌ Error downloading image from {url}: {e}")
        return None

# --- Part 2: Background Extraction Function (U2NETP Integration) ---

def extract_background_u2net(image_path: str, results_dir: str = "neww") -> str:
    """
    Extracts the background from an image using the U2NETP model.

    Args:
        image_path (str): The path to the input image.
        results_dir (str): Directory to save the extracted background image.

    Returns:
        str: The path to the saved background image, or None if extraction fails.
    """
    if not os.path.exists(image_path):
        print(f"❌ Error: Input image not found at {image_path}")
        return None

    # Configuration
    MODEL_PATH = "saved_models/u2net/u2netp.pth"
    MASK_SUFFIX = "_mask.png"
    BACKGROUND_SUFFIX = "_background.png"

    os.makedirs(results_dir, exist_ok=True)

    net = None
    try:
        # Load U2NETP model
        print("⏳ Loading U2NETP model...")
        net = U2NETP(3, 1)
        if not os.path.exists(MODEL_PATH):
            raise FileNotFoundError(f"U2NETP weights not found at {MODEL_PATH}. Skipping background extraction.")
        net.load_state_dict(torch.load(MODEL_PATH, map_location='cpu'))
        net.eval()
        print("✅ Model loaded.")
    except FileNotFoundError as e:
        print(f"❌ Error loading U2NETP model: {e}")
        return None
    except Exception as e:
        print(f"❌ An unexpected error occurred while loading U2NETP model: {e}")
        return None

    # Preprocessing pipeline
    transform = transforms.Compose([
        transforms.Resize((320, 320)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ])

    basename = os.path.splitext(os.path.basename(image_path))[0]
    
    try:
        image = Image.open(image_path).convert('RGB')
        original_size = image.size  # (W, H)

        # Transform and prepare input
        input_tensor = transform(image).unsqueeze(0)
        input_tensor = Variable(input_tensor)

        # Model inference
        with torch.no_grad():
            d1, *_ = net(input_tensor)
            pred = d1[:, 0, :, :].squeeze().cpu().numpy()
            pred = (pred - pred.min()) / (pred.max() - pred.min())
            pred = (pred * 255).astype(np.uint8)

        # Resize to original size and binarize
        mask = cv2.resize(pred, original_size)
        _, binary_mask = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)
        
        # Save the mask (optional, for debugging)
        mask_filename = os.path.join(results_dir, f"{basename}{MASK_SUFFIX}")
        cv2.imwrite(mask_filename, binary_mask)
        print(f"✅ Saved mask: {mask_filename}")

        # Load image again using OpenCV for BGR manipulation
        image_cv = cv2.imread(image_path)
        if image_cv is None:
            print(f"❌ Error: Could not load image with OpenCV at {image_path}")
            return None

        # Resize mask if necessary (should match image_cv dimensions)
        mask_cv = cv2.resize(binary_mask, (image_cv.shape[1], image_cv.shape[0]))

        # Invert the mask to get the background part
        inverted_mask = cv2.bitwise_not(mask_cv)

        # Extract the background by applying the inverted mask to the original image
        # Pixels where inverted_mask is 255 (white) will be kept from image_cv
        background_image = cv2.bitwise_and(image_cv, image_cv, mask=inverted_mask)
        
        # Optionally, if you want a purely white or black background where subject was,
        # you could use a white image and blend, but bitwise_and extracts original background.
        
        # Save the background image
        bg_output_path = os.path.join(results_dir, f"{basename}{BACKGROUND_SUFFIX}")
        cv2.imwrite(bg_output_path, background_image)
        print(f"✅ Background image saved: {bg_output_path}")
        return bg_output_path

    except Exception as e:
        print(f"❌ Error processing image for background extraction {image_path}: {e}")
        return None

# --- Part 3: Dominant Color Extraction Class ---
class ExtractColor():
    @staticmethod
    def safeSerialize(obj):
        default = lambda o: f"<<non-serializable: {type(o).__qualname__}>>"
        return json.dumps(obj, default=default)

    @staticmethod
    def colorToDf(input_tuple):
        """
        Converts the output of extcolors.extract_from_path to a pandas DataFrame.
        """
        if not isinstance(input_tuple, tuple) or len(input_tuple) != 2:
            print("Invalid input for colorToDf. Expected a tuple (colors_list, total_pixels).")
            return pd.DataFrame(columns=['c_code', 'occurence'])

        colors_list, _ = input_tuple # Ignore total_pixels for this part

        df_data = []
        for rgb_tuple, count in colors_list:
            hex_code = rgb2hex(rgb_tuple[0], rgb_tuple[1], rgb_tuple[2])
            df_data.append({'c_code': hex_code, 'occurence': count})

        df = pd.DataFrame(df_data)
        return df

    @staticmethod
    def hexToRgb(hex_code):
        hex_code = str(hex_code).strip('#')
        r = int(hex_code[0:2], 16)
        g = int(hex_code[2:4], 16)
        b = int(hex_code[4:6], 16)
        return r, g, b

    @staticmethod
    def is_single_tone(image_path, colors_x, threshold=5):
        """
        Checks if an image is predominantly a single tone.
        """
        # Ensure colors_x is a valid tuple (colors_list, total_pixels)
        if not isinstance(colors_x, tuple) or len(colors_x) != 2:
            print("Invalid input for is_single_tone. Expected a tuple (colors_list, total_pixels).")
            return "False"

        colors_list, total_pixels = colors_x
        if total_pixels == 0:
            return "False" # Avoid division by zero

        df_color = ExtractColor.colorToDf(colors_x)
        
        # Use the 'occurence' directly, as it's already a count from extcolors
        colors_count = list(df_color['occurence'])
        percentage_list = [int(c) * 100 / total_pixels for c in colors_count]
        percentage_list.sort()

        max_percentage = sum(percentage_list[-3:]) if len(percentage_list) >= 3 else sum(percentage_list)
        distinct_colors = len(df_color)

        if distinct_colors <= threshold or max_percentage >= 75:
            return "True"
        return "False"

    @staticmethod
    def extractColor(path):
        """
        Extracts the top 3 dominant colors from an image.
        Resizes image for faster processing if too large.
        """
        tolerance = 11

        img = Image.open(path)
        width, height = img.size
        
        # Resize image if width > 250 for faster color extraction
        if width > 250:
            new_width = 250
            new_height = int((height / width) * new_width)
            img = img.resize((new_width, new_height))
            # For simplicity, we'll process from the PIL Image object directly without saving
            # to disk again, as extcolors can take a PIL Image.
        
        colors_x = extcolors.extract_from_image(img, tolerance=tolerance, limit=13)
        df_color = ExtractColor.colorToDf(colors_x)

        top_3_colors = df_color['c_code'].head(3).tolist()
        return top_3_colors

# --- Part 4: Main Orchestration Function ---
def process_thumbnail_for_background_color(thumb_url: str):
    """
    Orchestrates the process of downloading a thumbnail, attempting background extraction,
    and then determining the dominant colors of the extracted background.

    Args:
        thumb_url (str): The URL of the thumbnail image.

    Returns:
        dict: A dictionary containing the original URL, paths to downloaded and background images,
              and the top dominant colors of the background, or error messages.
    """
    print(f"\n--- Starting processing for URL: {thumb_url} ---")
    results = {
        "original_url": thumb_url,
        "downloaded_image_path": None,
        "background_image_path": None,
        "dominant_background_colors": []
    }

    # Step 1: Download the thumbnail
    downloaded_path = download_image(thumb_url, output_dir="downloads")
    if not downloaded_path:
        results["status"] = "Failed to download image."
        return results
    results["downloaded_image_path"] = downloaded_path

    # Step 2: Extract background using U2NETP
    # Note: This step depends on 'model.py' and 'u2netp.pth' being available.
    background_path = extract_background_u2net(downloaded_path, results_dir="neww")
    if not background_path:
        results["status"] = "Background extraction failed or skipped due to missing model files."
        # If background extraction fails, we can still try to get colors from the original image
        # if the user's intent allows. For now, we'll indicate failure for background colors.
        # However, the prompt implies "dominant color from that image" (the extracted background).
        print("Attempting to extract dominant color from original image as a fallback...")
        try:
            top_colors = ExtractColor.extractColor(downloaded_path)
            results["dominant_background_colors"] = top_colors
            results["note"] = "Dominant colors extracted from original image as background extraction failed."
        except Exception as e:
            print(f"❌ Error extracting dominant color from original image: {e}")
            results["dominant_background_colors"] = []
            results["note"] = "Failed to extract any dominant colors."
        return results
    results["background_image_path"] = background_path

    # Step 3: Check dominant color from the extracted background image
    try:
        top_colors = ExtractColor.extractColor(background_path)
        results["dominant_background_colors"] = top_colors
        results["status"] = "Success"
    except Exception as e:
        print(f"❌ Error extracting dominant color from background image {background_path}: {e}")
        results["status"] = f"Failed to extract dominant colors: {e}"

    print(f"--- Finished processing for URL: {thumb_url} ---")
    return results

# --- Example Usage ---
if __name__ == "__main__":
    # Ensure U2NET model files are downloaded or available
    # The `u2netp.pth` still requires manual download and placement.
    model_files_ready = download_u2net_model_files()
    if not model_files_ready:
        print("\n🚫 Cannot proceed with full background extraction. Please manually download 'u2netp.pth' as instructed above.")
        # Exit or provide alternative processing if model not ready
        # For this example, we'll proceed, but background extraction will likely fail.

    # Example thumbnail URL. Replace with actual URLs from your 'thumb_url' column.
    # Note: Ensure this URL is accessible and points to an image.
    # For a real scenario, you would read your 'thumb_url' column from a DataFrame or CSV.
    
    sample_thumbnail_url = "https://stream.jdmagicbox.com/thumbnail/080.dko2gysrgwf/080.dko2gysrgwf_0enuhvjstexv8ui.jpg" 
    
    # Simulate data from a DataFrame 'thumb_url' column
    # For a real application, you would load your data, e.g.:
    # df = pd.read_csv('your_data.csv')
    # thumbnail_urls = df['thumb_url'].tolist()
    
    thumbnail_urls_to_process = [
        sample_thumbnail_url,
        # Add more URLs here to test a list of images
        # "https://example.com/another_image.jpg"
    ]

    all_results = []
    for url in thumbnail_urls_to_process:
        result = process_thumbnail_for_background_color(url)
        all_results.append(result)
        print(f"Result for {url}: {json.dumps(result, indent=2)}")

    print("\n--- All Processing Complete ---")
    print("Summary of results:")
    for res in all_results:
        print(f"URL: {res.get('original_url')}")
        print(f"  Status: {res.get('status')}")
        print(f"  Background Colors: {res.get('dominant_background_colors')}")
        if res.get('note'):
            print(f"  Note: {res.get('note')}")
        print("-" * 30)

```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
from torchvision import transforms
from torch.autograd import Variable
import requests
import pandas as pd
import json
from colormap import rgb2hex
import extcolors
import re
import io
import sys

# --- Configuration for U2NET Model Files ---
# Base directory where U2NET model files (model.py and saved_models) are located.
# This path is provided by the user.
U2NET_BASE_DIR = "/home/justdial/Videos/archive/timages"

# --- Function to check for U2NET model files ---
def check_u2net_model_files_existence():
    """
    Checks for the existence of U2NET model.py and u2netp.pth files
    within the specified U2NET_BASE_DIR.
    """
    MODEL_PY_PATH_EXPECTED = os.path.join(U2NET_BASE_DIR, "model", "model.py")
    U2NETP_PTH_PATH_EXPECTED = os.path.join(U2NET_BASE_DIR, "saved_models", "u2net", "u2netp.pth")
    
    print("\n--- Checking for U2NET model dependencies in specified directory ---")

    model_py_exists = os.path.exists(MODEL_PY_PATH_EXPECTED)
    u2netp_pth_exists = os.path.exists(U2NETP_PTH_PATH_EXPECTED)

    if not model_py_exists:
        print(f"❌ 'model.py' not found at: {MODEL_PY_PATH_EXPECTED}")
        print("ACTION REQUIRED: Please ensure 'model.py' is *exactly* named 'model.py' and located in the 'model' subdirectory within your specified U2NET base directory.")
        print(f"  Example structure: {U2NET_BASE_DIR}/model/model.py")
        print("  You might need to download it again from: https://github.com/xuebinqin/U-2-Net/blob/master/U-2-Net_static/model.py (click 'Raw' then 'Save Page As...')")
    else:
        print(f"✅ 'model.py' found at: {MODEL_PY_PATH_EXPECTED}")

    if not u2netp_pth_exists:
        print(f"❌ U2NETP pre-trained weights ('u2netp.pth') not found at: {U2NETP_PTH_PATH_EXPECTED}")
        print("ACTION REQUIRED: Please ensure 'u2netp.pth' is correctly placed in the 'saved_models/u2net/' subdirectory within your specified U2NET base directory.")
        print(f"  Example structure: {U2NET_BASE_DIR}/saved_models/u2net/u2netp.pth")
        print("  Download from U-2-Net releases (look for 'u2netp.pth'): https://github.com/xuebinqin/U-2-Net/releases")
    else:
        print(f"✅ 'u2netp.pth' found at: {U2NETP_PTH_PATH_EXPECTED}")

    # Return True only if both files are found
    if model_py_exists and u2netp_pth_exists:
        print("\nAll U2NET model dependencies found. Proceeding with background extraction.")
        return True
    else:
        print("\n🚫 Missing U2NET model dependencies. Full background extraction cannot proceed until both files are correctly placed.")
        return False

# IMPORTANT:
# The U2NETP model requires two external files:
# 1. model/model.py: Contains the U2NETP model architecture (class U2NETP).
# 2. saved_models/u2net/u2netp.pth: The pre-trained weights for the model.
# These files are NOT part of standard Python libraries and MUST BE PLACED MANUALLY
# in the directory structure relative to U2NET_BASE_DIR, as specified above.
# Ensure 'model.py' and 'u2netp.pth' are from compatible releases/versions to avoid 'Unexpected key(s)' errors.

# Dynamically add the 'model' subdirectory within U2NET_BASE_DIR to Python's path to import U2NETP.
MODEL_SUBDIR_PATH = os.path.join(U2NET_BASE_DIR, "model")
# Ensure the directory exists before attempting to add it to sys.path
if os.path.isdir(MODEL_SUBDIR_PATH) and MODEL_SUBDIR_PATH not in sys.path:
    sys.path.append(MODEL_SUBDIR_PATH)

try:
    from model import U2NETP
except ImportError:
    # This block is executed if `from model import U2NETP` fails.
    # This is the primary reason for your "Critical Error: Could not import U2NETP" message.
    print(f"❌ Critical Error: Could not import U2NETP from '{MODEL_SUBDIR_PATH}/model.py'.")
    print("This means 'model.py' is either missing, misnamed, or its directory is not correctly added to PATH.")
    print("Please verify the file path and name as instructed above.")
    class U2NETP(torch.nn.Module): # Fallback mock model to prevent immediate crash
        def __init__(self, in_ch=3, out_ch=1):
            super().__init__()
            print("U2NETP: Mock model loaded due to import failure. Background removal will not occur.")
        def forward(self, x):
            # Return dummy output if model not loaded, to allow script to run
            return torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3]), \
                   torch.zeros(x.shape[0], 1, x.shape[2], x.shape[3])


# --- Part 1: Download Image Function ---
def download_image(url: str, output_dir: str = "downloads", filename: str = None) -> str:
    """
    Downloads an image from a given URL and saves it to a specified directory.

    Args:
        url (str): The URL of the image to download.
        output_dir (str): The directory where the image will be saved. Defaults to "downloads".
        filename (str, optional): The name to save the file as. If None, derives from URL.

    Returns:
        str: The full path to the downloaded image, or None if download fails.
    """
    os.makedirs(output_dir, exist_ok=True)
    if filename is None:
        filename = url.split('/')[-1].split('?')[0] # Basic filename extraction
        if not filename: # Fallback if URL doesn't provide a good filename
            filename = "downloaded_image.png"
    
    file_path = os.path.join(output_dir, filename)

    try:
        print(f"📥 Downloading image from: {url}")
        response = requests.get(url, stream=True)
        response.raise_for_status() # Raise an exception for bad status codes

        with open(file_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"✅ Image downloaded to: {file_path}")
        return file_path
    except requests.exceptions.RequestException as e:
        print(f"❌ Error downloading image from {url}: {e}")
        return None

# --- Part 2: Background Extraction Function (U2NETP Integration) ---

def extract_background_u2net(image_path: str, results_dir: str = "results") -> str:
    """
    Extracts the background from an image using the U2NETP model.

    Args:
        image_path (str): The path to the input image.
        results_dir (str): Directory to save the extracted background image.

    Returns:
        str: The path to the saved background image, or None if extraction fails.
    """
    if not os.path.exists(image_path):
        print(f"❌ Error: Input image not found at {image_path}")
        return None

    # Configuration: MODEL_PATH now uses the U2NET_BASE_DIR
    MODEL_PATH = os.path.join(U2NET_BASE_DIR, "saved_models", "u2net", "u2netp.pth")
    MASK_SUFFIX = "_mask.png"
    BACKGROUND_SUFFIX = "_background.png"

    os.makedirs(results_dir, exist_ok=True)

    net = None
    try:
        # Load U2NETP model
        print("⏳ Loading U2NETP model...")
        net = U2NETP(3, 1) # This will be the actual U2NETP if import was successful
        if not os.path.exists(MODEL_PATH):
            raise FileNotFoundError(f"U2NETP weights not found at {MODEL_PATH}. Skipping background extraction.")
        
        # Load state_dict, allowing for some non-strictness if minor mismatches occur,
        # but the ideal is perfectly compatible model.py and .pth.
        net.load_state_dict(torch.load(MODEL_PATH, map_location='cpu'), strict=False)
        net.eval()
        print("✅ Model loaded.")
    except FileNotFoundError as e:
        print(f"❌ Error loading U2NETP model: {e}")
        return None
    except Exception as e:
        print(f"❌ An unexpected error occurred while loading U2NETP model: {e}")
        print("This often indicates an incompatibility between 'model.py' and 'u2netp.pth'.")
        print("Please ensure both files are from the *same release* of the U-2-Net project.")
        return None

    # Preprocessing pipeline
    transform = transforms.Compose([
        transforms.Resize((320, 320)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ])

    basename = os.path.splitext(os.path.basename(image_path))[0]
    
    try:
        image = Image.open(image_path).convert('RGB')
        original_size = image.size  # (W, H)

        # Transform and prepare input
        input_tensor = transform(image).unsqueeze(0)
        input_tensor = Variable(input_tensor)

        # Model inference
        with torch.no_grad():
            d1, *_ = net(input_tensor)
            pred = d1[:, 0, :, :].squeeze().cpu().numpy()
            pred = (pred - pred.min()) / (pred.max() - pred.min())
            pred = (pred * 255).astype(np.uint8)

        # Resize to original size and binarize
        mask = cv2.resize(pred, original_size)
        _, binary_mask = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)
        
        # Save the mask (optional, for debugging)
        mask_filename = os.path.join(results_dir, f"{basename}{MASK_SUFFIX}")
        cv2.imwrite(mask_filename, binary_mask)
        print(f"✅ Saved mask: {mask_filename}")

        # Load image again using OpenCV for BGR manipulation
        image_cv = cv2.imread(image_path)
        if image_cv is None:
            print(f"❌ Error: Could not load image with OpenCV at {image_path}")
            return None

        # Resize mask if necessary (should match image_cv dimensions)
        mask_cv = cv2.resize(binary_mask, (image_cv.shape[1], image_cv.shape[0]))

        # Invert the mask to get the background part
        inverted_mask = cv2.bitwise_not(mask_cv)

        # Extract the background by applying the inverted mask to the original image
        # Pixels where inverted_mask is 255 (white) will be kept from image_cv
        background_image = cv2.bitwise_and(image_cv, image_cv, mask=inverted_mask)
        
        # Optionally, if you want a purely white or black background where subject was,
        # you could use a white image and blend, but bitwise_and extracts original background.
        
        # Save the background image
        bg_output_path = os.path.join(results_dir, f"{basename}{BACKGROUND_SUFFIX}")
        cv2.imwrite(bg_output_path, background_image)
        print(f"✅ Background image saved: {bg_output_path}")
        return bg_output_path

    except Exception as e:
        print(f"❌ Error processing image for background extraction {image_path}: {e}")
        return None

# --- Part 3: Dominant Color Extraction Class ---
class ExtractColor():
    @staticmethod
    def safeSerialize(obj):
        default = lambda o: f"<<non-serializable: {type(o).__qualname__}>>"
        return json.dumps(obj, default=default)

    @staticmethod
    def colorToDf(input_tuple):
        """
        Converts the output of extcolors.extract_from_path to a pandas DataFrame.
        """
        if not isinstance(input_tuple, tuple) or len(input_tuple) != 2:
            print("Invalid input for colorToDf. Expected a tuple (colors_list, total_pixels).")
            return pd.DataFrame(columns=['c_code', 'occurence'])

        colors_list, _ = input_tuple # Ignore total_pixels for this part

        df_data = []
        for rgb_tuple, count in colors_list:
            hex_code = rgb2hex(rgb_tuple[0], rgb_tuple[1], rgb_tuple[2])
            df_data.append({'c_code': hex_code, 'occurence': count})

        df = pd.DataFrame(df_data)
        return df

    @staticmethod
    def hexToRgb(hex_code):
        hex_code = str(hex_code).strip('#')
        r = int(hex_code[0:2], 16)
        g = int(hex_code[2:4], 16)
        b = int(hex_code[4:6], 16)
        return r, g, b

    @staticmethod
    def is_single_tone(image_path, colors_x, threshold=5):
        """
        Checks if an image is predominantly a single tone.
        """
        # Ensure colors_x is a valid tuple (colors_list, total_pixels)
        if not isinstance(colors_x, tuple) or len(colors_x) != 2:
            print("Invalid input for is_single_tone. Expected a tuple (colors_list, total_pixels).")
            return "False"

        colors_list, total_pixels = colors_x
        if total_pixels == 0:
            return "False" # Avoid division by zero

        df_color = ExtractColor.colorToDf(colors_x)
        
        # Use the 'occurence' directly, as it's already a count from extcolors
        colors_count = list(df_color['occurence'])
        percentage_list = [int(c) * 100 / total_pixels for c in colors_count]
        percentage_list.sort()

        max_percentage = sum(percentage_list[-3:]) if len(percentage_list) >= 3 else sum(percentage_list)
        distinct_colors = len(df_color)

        if distinct_colors <= threshold or max_percentage >= 75:
            return "True"
        return "False"

    @staticmethod
    def extractColor(path):
        """
        Extracts the top 3 dominant colors from an image.
        Resizes image for faster processing if too large.
        """
        tolerance = 11

        img = Image.open(path)
        width, height = img.size
        
        # Resize image if width > 250 for faster color extraction
        if width > 250:
            new_width = 250
            new_height = int((height / width) * new_width)
            img = img.resize((new_width, new_height))
            # For simplicity, we'll process from the PIL Image object directly without saving
            # to disk again, as extcolors can take a PIL Image.
        
        colors_x = extcolors.extract_from_image(img, tolerance=tolerance, limit=13)
        df_color = ExtractColor.colorToDf(colors_x)

        top_3_colors = df_color['c_code'].head(3).tolist()
        return top_3_colors

# --- Part 4: Main Orchestration Function ---
def process_thumbnail_for_background_color(thumb_url: str):
    """
    Orchestrates the process of downloading a thumbnail, attempting background extraction,
    and then determining the dominant colors of the extracted background.

    Args:
        thumb_url (str): The URL of the thumbnail image.

    Returns:
        dict: A dictionary containing the original URL, paths to downloaded and background images,
              and the top dominant colors of the background, or error messages.
    """
    print(f"\n--- Starting processing for URL: {thumb_url} ---")
    results = {
        "original_url": thumb_url,
        "downloaded_image_path": None,
        "background_image_path": None,
        "dominant_background_colors": []
    }

    # Step 1: Download the thumbnail
    downloaded_path = download_image(thumb_url, output_dir="downloads")
    if not downloaded_path:
        results["status"] = "Failed to download image."
        return results
    results["downloaded_image_path"] = downloaded_path

    # Step 2: Extract background using U2NETP
    # Note: This step depends on 'model.py' and 'u2netp.pth' being available.
    background_path = extract_background_u2net(downloaded_path, results_dir="results")
    if not background_path:
        results["status"] = "Background extraction failed or skipped due to missing model files."
        # If background extraction fails, we can still try to get colors from the original image
        # if the user's intent allows. For now, we'll indicate failure for background colors.
        # However, the prompt implies "dominant color from that image" (the extracted background).
        print("Attempting to extract dominant color from original image as a fallback...")
        try:
            top_colors = ExtractColor.extractColor(downloaded_path)
            results["dominant_background_colors"] = top_colors
            results["note"] = "Dominant colors extracted from original image as background extraction failed."
        except Exception as e:
            print(f"❌ Error extracting dominant color from original image: {e}")
            results["dominant_background_colors"] = []
            results["note"] = "Failed to extract any dominant colors."
        return results
    results["background_image_path"] = background_path

    # Step 3: Check dominant color from the extracted background image
    try:
        top_colors = ExtractColor.extractColor(background_path)
        results["dominant_background_colors"] = top_colors
        results["status"] = "Success"
    except Exception as e:
        print(f"❌ Error extracting dominant color from background image {background_path}: {e}")
        results["status"] = f"Failed to extract dominant colors: {e}"

    print(f"--- Finished processing for URL: {thumb_url} ---")
    return results

# --- Example Usage ---
if __name__ == "__main__":
    # Ensure U2NET model files are downloaded or available at the specified path.
    model_files_ready = check_u2net_model_files_existence()
    if not model_files_ready:
        print("\n🚫 Cannot proceed with full background extraction without all model files in the specified directory.")
        # We will still attempt to process images, but background extraction will fall back to original image color extraction.
        # It's recommended to manually get the files for full functionality.

    # Example thumbnail URL. Replace with actual URLs from your 'thumb_url' column.
    # Note: Ensure this URL is accessible and points to an image.
    # For a real scenario, you would read your 'thumb_url' column from a DataFrame or CSV.
    
    sample_thumbnail_url = "https://stream.jdmagicbox.com/thumbnail/080.djof0cki5pf/080.djof0cki5pf_dyhbaiivyk0babe.jpg" 
    
    # Simulate data from a DataFrame 'thumb_url' column
    # For a real application, you would load your data, e.g.:
    # df = pd.read_csv('your_data.csv')
    # thumbnail_urls = df['thumb_url'].tolist()
    
    thumbnail_urls_to_process = [
        sample_thumbnail_url,
        # Add more URLs here to test a list of images
        # "https://example.com/another_image.jpg"
    ]

    all_results = []
    for url in thumbnail_urls_to_process:
        result = process_thumbnail_for_background_color(url)
        all_results.append(result)
        print(f"Result for {url}: {json.dumps(result, indent=2)}")

    print("\n--- All Processing Complete ---")
    print("Summary of results:")
    for res in all_results:
        print(f"URL: {res.get('original_url')}")
        print(f"  Status: {res.get('status')}")
        print(f"  Background Colors: {res.get('dominant_background_colors')}")
        if res.get('note'):
            print(f"  Note: {res.get('note')}")
        print("-" * 30)
```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
from torchvision import transforms
from torch.autograd import Variable

# --- START: Debugging File Paths ---
print("--- Checking file paths ---")
current_dir = os.getcwd()
model_dir_path = os.path.join(current_dir, 'model')
u2net_file_path = os.path.join(model_dir_path, 'u2net.py')

print(f"Current working directory: {current_dir}")

if os.path.exists(model_dir_path):
    print(f"✅ 'model/' directory found at: {model_dir_path}")
    if os.path.isdir(model_dir_path):
        print(f"Contents of 'model/' directory: {os.listdir(model_dir_path)}")
        if os.path.exists(u2net_file_path):
            print(f"✅ 'u2net.py' file found at: {u2net_file_path}")
        else:
            print(f"❌ 'u2net.py' NOT found inside 'model/' directory. Please ensure it's named 'u2net.py' and placed correctly.")
            print("Expected path: " + u2net_file_path)
            # You might want to exit here if the file is critical
            # exit()
    else:
        print(f"❌ 'model' exists but is not a directory. Please check your file system.")
        # exit()
else:
    print(f"❌ 'model/' directory NOT found at: {model_dir_path}. Please create this directory.")
    # exit()
print("--- Path check complete ---")
# --- END: Debugging File Paths ---

# Import your U2NETP model definition from the correct path: model/u2net.py
# Ensure that 'model' is a directory in the same location as this script,
# and 'u2net.py' is inside that 'model' directory.
from model.u2net import U2NETP # Corrected import path

# Configuration
# Make sure this path is correct relative to where you run the script.
# If 'saved_models' is at the same level as your script and 'model' folder,
# then 'saved_models/u2net/u2netp.pth' is correct.
MODEL_PATH = "saved_models/u2net/u2netp.pth"
RESULTS_DIR = "results"
MASK_SUFFIX = "_mask.png"

# Create results directory if it doesn't exist
os.makedirs(RESULTS_DIR, exist_ok=True)

# Load U2NETP model
print("⏳ Loading U2NETP model...")
# Initialize the model with 3 input channels (RGB image) and 1 output channel (mask)
net = U2NETP(3, 1)

# Load the pre-trained weights.
# 'map_location="cpu"' ensures it loads correctly even if you don't have a GPU.
try:
    net.load_state_dict(torch.load(MODEL_PATH, map_location='cpu'))
    print("✅ Model weights loaded successfully.")
except FileNotFoundError:
    print(f"❌ Error: Model weights file not found at '{MODEL_PATH}'.")
    print("Please ensure the 'saved_models/u2net/u2netp.pth' file exists in the correct path.")
    exit() # Exit the script if the model weights are not found
except Exception as e:
    print(f"❌ An error occurred while loading model weights: {e}")
    exit()

# Set the model to evaluation mode (disables dropout, batchnorm updates, etc.)
net.eval()
print("✅ Model is in evaluation mode.")

# Define the preprocessing pipeline for input images
transform = transforms.Compose([
    transforms.Resize((320, 320)), # Resize image to 320x320, as expected by U2NETP
    transforms.ToTensor(),         # Convert PIL Image to PyTorch Tensor (scales to [0, 1])
    transforms.Normalize(          # Normalize with ImageNet mean and std dev
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("\nStarting image processing...")

# Process each image in the current directory
# Adjust this loop if your images are in a different folder (e.g., 'images/')
supported_exts = ('.jpg', '.jpeg', '.png')
for filename in os.listdir('.'): # Iterates through files in the current directory
    # Skip files that are not supported image types
    if not filename.lower().endswith(supported_exts):
        continue

    # Skip files that might be the mask output from previous runs or the model itself
    # Also skip the model files themselves from being processed as images
    if filename.endswith(MASK_SUFFIX) or filename.endswith("_subject.png") or filename == "u2net.py" or filename == "model.py":
        continue
    # If the file is inside the 'model' directory, skip it.
    if os.path.dirname(filename) == 'model':
        continue


    print(f"\n🔍 Processing image: {filename}")
    basename = os.path.splitext(filename)[0] # Get filename without extension

    try:
        # Open the image using PIL and convert to RGB (important for consistent 3 channels)
        image = Image.open(filename).convert('RGB')
    except Exception as e:
        print(f"❌ Could not open image {filename}: {e}. Skipping.")
        continue

    original_size = image.size  # (Width, Height) of the original image

    # Apply transformations and add a batch dimension (unsqueeze(0))
    input_tensor = transform(image).unsqueeze(0)
    # Wrap in Variable for older PyTorch versions, though not strictly necessary in newer ones
    input_tensor = Variable(input_tensor)

    # Perform model inference
    with torch.no_grad(): # Disable gradient calculation for inference to save memory and speed up
        d1, *_ = net(input_tensor) # U2NETP returns multiple side outputs, we only need the first (d1)
        # Process the output:
        # 1. Squeeze to remove batch and channel dimensions
        # 2. Move to CPU and convert to NumPy array
        pred = d1[:, 0, :, :].squeeze().cpu().numpy()

        # Normalize prediction to 0-1 range (min-max normalization)
        pred = (pred - pred.min()) / (pred.max() - pred.min() + 1e-8) # Add small epsilon to prevent division by zero
        # Scale to 0-255 and convert to unsigned 8-bit integer for image saving
        pred = (pred * 255).astype(np.uint8)

    # Resize the predicted mask to the original image dimensions
    mask = cv2.resize(pred, original_size)
    # Binarize the mask: pixels > 127 become 255 (white), others 0 (black)
    _, binary_mask = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)

    # Save the binary mask
    mask_filename = os.path.join(RESULTS_DIR, f"{basename}{MASK_SUFFIX}") # Save mask in results directory
    cv2.imwrite(mask_filename, binary_mask)
    print(f"✅ Saved mask: {mask_filename}")

    # Load original image and the generated mask using OpenCV for BGRA manipulation
    # imread loads in BGR format by default
    image_cv = cv2.imread(filename)
    mask_cv = cv2.imread(mask_filename, 0) # Load mask as grayscale

    # Ensure mask has the same dimensions as the image_cv if not already
    if image_cv.shape[0] != mask_cv.shape[0] or image_cv.shape[1] != mask_cv.shape[1]:
        mask_cv = cv2.resize(mask_cv, (image_cv.shape[1], image_cv.shape[0]))

    # Split the BGR channels from the image
    b, g, r = cv2.split(image_cv)
    # Merge B, G, R, and the mask (as alpha channel) to create an RGBA image
    rgba = cv2.merge((b, g, r, mask_cv))

    # Save the transparent subject image
    output_path = os.path.join(RESULTS_DIR, f"{basename}_subject.png")
    cv2.imwrite(output_path, rgba)
    print(f"✅ Transparent subject image saved: {output_path}")

print("\n✨ All images processed. Check the 'results' directory for outputs.")
```

```python
import os
import pandas as pd
import numpy as np # Keep import as per original structure, though not directly used in the modified logic
import json # Keep import as per original structure, though not directly used
from colormap import rgb2hex # Re-added as per user's structural requirement
import extcolors # Re-added as per user's structural requirement
from PIL import Image

class ExtractColor():
    """
    A utility class to extract color information from images.
    Adheres to the user's provided structure using 'colormap' and 'extcolors'.
    """

    @staticmethod
    def colorToDf(input_colors_list):
        """
        Converts the raw output from extcolors.extract_from_image into a pandas DataFrame.
        This method directly processes the list of ((R, G, B), count) tuples.

        Args:
            input_colors_list (list): A list of tuples, where each tuple is
                                      ((R, G, B), count), as returned by extcolors.

        Returns:
            pandas.DataFrame: A DataFrame with 'c_code' (hex color) and 'occurence' columns.
        """
        df_color_up = []
        df_percent = []

        # Iterate directly over the list of color tuples and their counts
        for rgb_tuple, count in input_colors_list:
            # Convert RGB tuple to hex color code using rgb2hex from 'colormap'
            hex_code = rgb2hex(rgb_tuple[0], rgb_tuple[1], rgb_tuple[2])
            df_color_up.append(hex_code)
            # Store occurrence as a string as per original 'df_percent' behavior
            df_percent.append(str(count))

        # Create DataFrame from hex codes and occurrences
        df = pd.DataFrame(zip(df_color_up, df_percent),
                            columns=['c_code', 'occurence'])
        return df

    @staticmethod
    def extractColor(path):
        """
        Extracts the second most dominant color from an image file after identifying the top three.
        This method uses 'extcolors' for color extraction and 'colormap' for hex conversion.
        It processes the image in-memory after resizing, avoiding disk I/O for resized images.

        Args:
            path (str): The file path to the image.

        Returns:
            str: The hexadecimal color code of the second dominant color, or None if not found.
        """
        tolerance = 11 # Tolerance level for grouping similar colors
        img = None # Initialize img variable

        try:
            img = Image.open(path)
            # Ensure image is in RGB format for consistent pixel data and extcolors compatibility
            img = img.convert('RGB')

            width, height = img.size
            # Resize image in-memory if its width is greater than 250 pixels, maintaining aspect ratio.
            if width > 250:
                new_width = 250
                new_height = int((height / width) * new_width)
                # Use LANCZOS for high-quality downsampling during resizing
                img = img.resize((new_width, new_height), Image.Resampling.LANCZOS)
                # Removed img.convert('RGB').save(path) - no longer saving resized image to disk

            # Extract colors using extcolors.extract_from_image, which accepts a PIL Image object.
            # We limit to 3 to get the top three dominant colors.
            # extcolors returns a tuple: (list_of_colors, total_pixel_count). We only need the list of colors.
            colors_list_from_extcolors, _ = extcolors.extract_from_image(img, tolerance=tolerance, limit=3) # Changed limit to 3

            # Convert the extracted colors list to a DataFrame using the refactored method
            df_color = ExtractColor.colorToDf(colors_list_from_extcolors)

            # Return the second dominant color (index 1) if available, otherwise None.
            if len(df_color) > 1: # Check if at least two colors were extracted
                return df_color['c_code'][1] # Return the second dominant color
            elif not df_color.empty: # If only one color was extracted, return it as a fallback
                return df_color['c_code'][0]
            return None # Return None if no colors were extracted or DataFrame is empty
        except Exception as e:
            # Print the specific error that occurred during color extraction for debugging
            print(f"❌ Error during color extraction for {path}: {e}")
            return None

# ========================== Pipeline Execution ==========================

if __name__ == "__main__":
    # Define folder paths and file names based on user's request
    image_extensions = ('.png',) # Only process PNG files
    new_results_dir = "new_results_sriram" # User specified image folder
    # Changed excel_path to point to the new CSV file
    excel_path = "/home/justdial/Videos/archive/Master Data_GoJD.csv"
    output_excel_path = "Master Data_GoJD_with_dominant_color_latest.xlsx" # New output Excel file name

    # ----------------------------------------------------------------------
    # Step 1: Check if the image directory exists.
    # ----------------------------------------------------------------------
    if not os.path.exists(new_results_dir):
        print(f"❌ Folder '{new_results_dir}' not found. Please ensure it exists.")
        exit() # Terminate script if folder is missing

    # ----------------------------------------------------------------------
    # Step 2: Find all '_subject.png' files in the specified directory.
    # ----------------------------------------------------------------------
    files = [f for f in os.listdir(new_results_dir) if f.lower().endswith('_subject.png')]

    if not files:
        print(f"❌ No '_subject.png' images found in '{new_results_dir}'.")
        exit() # Terminate script if no relevant images are found

    print(f"🔍 Found {len(files)} subject images to process for dominant color extraction...")

    # ----------------------------------------------------------------------
    # Step 3: Load the Master Data file (now CSV).
    # ----------------------------------------------------------------------
    df = pd.DataFrame() # Initialize an empty DataFrame to handle loading errors gracefully
    try:
        # Changed to pd.read_csv for CSV file support
        df = pd.read_csv(excel_path)
        print(f"✅ CSV loaded successfully: {excel_path}")
    except FileNotFoundError:
        print(f"❌ Error: CSV file '{excel_path}' not found. Please verify the path.")
        exit() # Terminate script if CSV file is missing
    except Exception as e:
        print(f"❌ Error loading CSV: {e}")
        exit() # Terminate script for other loading errors

    # ----------------------------------------------------------------------
    # Step 4: Add the 'dominant_color' column if it doesn't exist.
    # ----------------------------------------------------------------------
    if 'dominant_color' not in df.columns:
        df['dominant_color'] = ''
        print("📝 Added 'dominant_color' column to the DataFrame.")

    # Check: Ensure 'product_thumb' column exists
    if 'product_thumb' not in df.columns:
        print(f"❌ Error: Required column 'product_thumb' not found in the CSV file. Please ensure it exists.")
        exit() # Exit if the critical column is missing

    # Extract base filename from 'product_thumb' URLs once
    # This creates a temporary column in the DataFrame for efficient matching.
    df['product_thumb_base_name'] = df['product_thumb'].astype(str).apply(
        lambda x: os.path.splitext(os.path.basename(x.split('/')[-1]))[0] if pd.notna(x) and '/' in x else ''
    )

    # ----------------------------------------------------------------------
    # Step 5: Process each image, extract the dominant color, and update the DataFrame.
    # ----------------------------------------------------------------------
    for image_file in files:
        try:
            image_path = os.path.join(new_results_dir, image_file)
            # The extractColor method will now return the second dominant color
            dominant_color = ExtractColor.extractColor(image_path)
            
            if dominant_color:
                print(f"{image_file} ➤ Storing First Dominant Color: {dominant_color}")

                # Get the base filename from the subject image file (e.g., '-q29heqdl4c')
                base_filename = image_file.replace('_subject.png', '')

                # Find matching row(s) in the DataFrame using the new 'product_thumb_base_name' column.
                # This uses an exact match to ensure correct linking.
                matched_idx = df[df['product_thumb_base_name'] == base_filename].index

                if not matched_idx.empty:
                    # Update the 'dominant_color' column for all found matching indices.
                    for idx in matched_idx:
                        df.at[idx, 'dominant_color'] = dominant_color
                    print(f"✅ Updated dominant_color for entries matching: {base_filename}")
                else:
                    print(f"⚠️ No match found in CSV for: {base_filename}") # Changed message for CSV
            else:
                print(f"❌ Could not extract sufficient dominant colors from {image_file}.")

        except Exception as e:
            print(f"❌ An unexpected error occurred while processing {image_file} in the main loop: {e}")

    # ----------------------------------------------------------------------
    # Step 6: Save the updated DataFrame to a new Excel file.
    # ----------------------------------------------------------------------
    if not df.empty:
        try:
            # Save to a new .xlsx file (pandas defaults to openpyxl for .xlsx)
            df.to_excel(output_excel_path, index=False)
            print(f"\n✅ Excel successfully updated with dominant colors and saved to: {output_excel_path}")
        except Exception as e:
            print(f"❌ Error saving Excel: {e}")
    else:
        print("No DataFrame to save. It might be due to an error during loading the CSV file.")
```

```python
import pandas as pd
import numpy as np
import json
from flask import Flask # Included for consistency with original snippet, though not used in the class itself
from colormap import rgb2hex
import extcolors
from PIL import Image
import re
import os
import cv2
import torch
from torchvision import transforms
from torch.autograd import Variable

# --- PLACEHOLDER FOR U2NETP MODEL DEFINITION ---
# IMPORTANT: You MUST replace this with your actual U2NETP model architecture
# from your 'model.py' file for the background removal to function correctly.
class U2NETP(torch.nn.Module):
    def __init__(self, in_ch=3, out_ch=1):
        super(U2NETP, self).__init__()
        # This is a minimalistic placeholder. Your actual U2NETP
        # implementation will have many more layers.
        self.conv = torch.nn.Conv2d(in_ch, out_ch, kernel_size=1)

    def forward(self, x):
        # This is a placeholder forward pass.
        # Your actual U2NETP forward pass should return a tuple of predictions,
        # typically d1, d2, d3, d4, d5, d6, d7.
        # For this example, we'll return d1 and fill the rest with None.
        return [self.conv(x), None, None, None, None, None, None]

# --- END OF PLACEHOLDER ---


class ExtractColor():
    def __init__(self):
        """
        Initializes the ExtractColor class.
        Currently, no heavy initialization (like model loading) is done here
        to allow the remove_background method to be self-contained for model loading.
        For performance in repeated calls, consider loading the U2NETP model in __init__
        if you reuse the class instance.
        """
        pass

    @staticmethod
    def safeSerialize(obj):
        """
        Safely serializes an object to JSON, replacing non-serializable objects
        with a descriptive string.
        """
        default = lambda o: f"<<non-serializable: {type(o).__qualname__}>>"
        return json.dumps(obj, default=default)

    @staticmethod
    def colorToDf(input_colors_data):
        """
        Converts the output from extcolors.extract_from_path to a pandas DataFrame
        with hex codes and occurrence percentages.

        Args:
            input_colors_data (tuple): Tuple containing (list of (RGB, count) tuples, total_pixels).

        Returns:
            pd.DataFrame: DataFrame with 'c_code' (hex) and 'occurence' (pixel count).
        """
        colors_list, _ = input_colors_data
        
        df_rgb = [f"({r}, {g}, {b})" for (r, g, b), _ in colors_list]
        df_percent = [count for (_, count) in colors_list]

        # Convert RGB to HEX code
        df_color_up = [rgb2hex(rgb[0], rgb[1], rgb[2]) for rgb, _ in colors_list]

        df = pd.DataFrame(zip(df_color_up, df_percent),
                            columns=['c_code', 'occurence'])
        return df

    @staticmethod
    def hexToRgb(hex_code):
        """
        Converts a hexadecimal color code to an RGB tuple.

        Args:
            hex_code (str): The hexadecimal color code (e.g., "#RRGGBB" or "RRGGBB").

        Returns:
            tuple: A tuple (r, g, b) representing the RGB values.
        """
        hex_code = str(hex_code).strip('#')  # Remove '#' symbol
        r = int(hex_code[0:2], 16)  # Convert red component from hex to decimal
        g = int(hex_code[2:4], 16)  # Convert green component from hex to decimal
        b = int(hex_code[4:6], 16)  # Convert blue component from hex to decimal
        return r, g, b

    @staticmethod
    def is_single_tone(image_path, colors_x, threshold=5):
        """
        Determines if an image is predominantly a single tone based on distinct colors
        and the percentage of the most dominant color(s).

        Args:
            image_path (str): Path to the image (used for context, not re-read).
            colors_x (tuple): Output from extcolors.extract_from_path.
            threshold (int): Maximum number of distinct colors to consider as single tone.

        Returns:
            str: "True" if single tone, "False" otherwise.
        """
        _, total_pixels = colors_x
        df_color = ExtractColor.colorToDf(colors_x)
        colors_count = list(df_color['occurence'])

        percentage_list = []
        for color_pixels in colors_count:
            color_pixels = int(color_pixels)
            color_percentage = color_pixels * 100 / total_pixels
            percentage_list.append(color_percentage)

        max_percentage = 0
        percentage_list.sort()

        # The original logic for max_percentage. It effectively takes only the last (highest) percentage
        # if the list is not empty, due to the structure of if/elif.
        if len(percentage_list) >= 1:
            max_percentage = percentage_list[-1]
        # The subsequent 'elif' checks would never be true if the first 'if' was true.
        # To sum top N percentages, a different loop or slice would be needed.
        # Example for summing top 3:
        # if len(percentage_list) > 0:
        #     max_percentage = sum(percentage_list[-min(3, len(percentage_list)):])


        distinct_colors = len(df_color)

        if distinct_colors <= threshold:
            return "True"
        elif (max_percentage >= 75):
            return "True"
        else:
            return "False"

    @staticmethod
    def extractColor(path):
        """
        Extracts dominant colors from an image, optionally resizing it,
        and determines if it's single-toned.

        Args:
            path (str): Path to the input image.

        Returns:
            tuple: A tuple containing:
                - list: A list of dictionaries, each describing a color (hex, percentage, rgb_values).
                - str: "True" or "False" indicating if the image is single-toned.
                - str: The path to the (potentially resized and overwritten) image.
        """
        result_rows = []
        tolerance = 11

        try:
            img = Image.open(path)
        except FileNotFoundError:
            print(f"Error: Image file not found at {path}")
            return [], "False", path
        except Exception as e:
            print(f"Error opening image {path}: {e}")
            return [], "False", path

        width = img.width
        height = img.height
        if width > 250:
            new_width = 250
            new_height = int((height / width) * new_width)

            img = img.resize((new_width, new_height), Image.Resampling.LANCZOS)
            img.convert('RGB').save(path) # Overwrite original with resized image

        colors_x = extcolors.extract_from_path(path, tolerance=tolerance, limit=13)
        df_color = ExtractColor.colorToDf(colors_x)

        list_color = list(df_color['c_code'])
        rgb_values = [ExtractColor.hexToRgb(hex_code) for hex_code in list_color]
        list_precent = [int(i) for i in list(df_color['occurence'])]

        # Calculate percentages safely to avoid division by zero if sum is 0
        total_pixels_sum = sum(list_precent)
        if total_pixels_sum == 0:
            text_c = []
        else:
            text_c = [{'hex': str(c), 'percentage': round(p * 100 / total_pixels_sum, 1), 'rbg_values': str(r)}
                      for c, p, r in zip(list_color, list_precent, rgb_values)]

        result_row = text_c
        result_rows.append(result_row)

        output_dict = result_rows
        rearranged_dict = output_dict[0]

        verdict = ExtractColor.is_single_tone(path, colors_x)

        return rearranged_dict, verdict, path

    def remove_background(self, image_path, model_path="saved_models/u2net/u2netp.pth", results_dir="results"):
        """
        Removes the background from a single image using a U2NETP model and saves the transparent subject.

        Args:
            image_path (str): Path to the input image file.
            model_path (str): Path to the U2NETP model weights (e.g., 'saved_models/u2net/u2netp.pth').
            results_dir (str): Directory where the output transparent image will be saved.

        Returns:
            str: Path to the saved transparent subject image, or None if an error occurred.
        """
        try:
            # Create results directory if it doesn't exist
            os.makedirs(results_dir, exist_ok=True)

            # Load U2NETP model
            print("⏳ Loading U2NETP model...")
            net = U2NETP(3, 1) # Ensure this U2NETP is your actual model definition
            
            # Check if model path exists
            if not os.path.exists(model_path):
                print(f"Error: Model file not found at {model_path}. Please provide a valid path.")
                return None

            net.load_state_dict(torch.load(model_path, map_location='cpu'))
            net.eval() # Set model to evaluation mode
            print("✅ Model loaded.")

            # Preprocessing pipeline
            transform = transforms.Compose([
                transforms.Resize((320, 320)),
                transforms.ToTensor(),
                transforms.Normalize([0.485, 0.456, 0.406],
                                     [0.229, 0.224, 0.225])
            ])

            print(f"\n🔍 Processing: {image_path}")
            # Get base filename without extension
            basename = os.path.splitext(os.path.basename(image_path))[0]
            
            # Open image using Pillow
            image = Image.open(image_path).convert('RGB')
            original_size = image.size  # (Width, Height)

            # Transform and prepare input tensor for the model
            input_tensor = transform(image).unsqueeze(0) # Add batch dimension
            input_tensor = Variable(input_tensor) # Wrap in Variable for older torch versions if needed

            # Model inference: get the saliency map (d1)
            with torch.no_grad(): # Disable gradient calculation for inference
                d1, *_ = net(input_tensor) # U2NET typically returns d1-d7, we need d1
                
                # Process the saliency map to a binary mask
                pred = d1[:, 0, :, :].squeeze().cpu().numpy() # Get the first channel, remove batch/channel dims, move to CPU, convert to numpy
                pred = (pred - pred.min()) / (pred.max() - pred.min()) # Normalize to 0-1
                pred = (pred * 255).astype(np.uint8) # Scale to 0-255

            # Resize the predicted mask to the original image dimensions
            mask = cv2.resize(pred, original_size, interpolation=cv2.INTER_LINEAR)
            
            # Apply binary threshold to create a crisp mask (0 or 255)
            _, binary_mask = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)
            
            # Save the binary mask temporarily (optional, can be skipped if not needed for debugging)
            mask_filename_temp = os.path.join(results_dir, f"{basename}_temp_mask.png")
            cv2.imwrite(mask_filename_temp, binary_mask)
            print(f"✅ Saved temporary mask: {mask_filename_temp}")

            # Load the original image and the generated mask using OpenCV
            image_cv = cv2.imread(image_path)
            mask_cv = cv2.imread(mask_filename_temp, cv2.IMREAD_GRAYSCALE) # Load mask as grayscale

            # Ensure mask is the same size as the image (should be from resize above)
            mask_cv = cv2.resize(mask_cv, (image_cv.shape[1], image_cv.shape[0]), interpolation=cv2.INTER_LINEAR)

            # Split the image into B, G, R channels
            b, g, r = cv2.split(image_cv)
            # Merge B, G, R with the mask as the alpha (transparency) channel
            rgba = cv2.merge((b, g, r, mask_cv))

            # Define the output path for the transparent subject image
            output_path = os.path.join(results_dir, f"{basename}_subject.png")
            cv2.imwrite(output_path, rgba) # Save the RGBA image
            print(f"✅ Transparent subject image saved: {output_path}")

            # Clean up the temporary mask file
            os.remove(mask_filename_temp)
            print(f"🧹 Cleaned up temporary mask: {mask_filename_temp}")

            return output_path

        except FileNotFoundError:
            print(f"❌ Error: Image file not found at {image_path}")
            return None
        except Exception as e:
            print(f"❌ An error occurred during background removal: {e}")
            return None


# Example Usage:
if __name__ == "__main__":
    # Create a dummy image for demonstration
    from PIL import Image, ImageDraw
    dummy_image_path = "/home/justdial/Videos/archive/timages/results1/080.c-aaqsqvxcc_zp2yzqkcjwoyehc_subject.png"
    img = Image.new('RGB', (100, 100), color = (255, 0, 0)) # Red background
    draw = ImageDraw.Draw(img)
    draw.ellipse((25, 25, 75, 75), fill=(0, 0, 255)) # Blue circle
    img.save(dummy_image_path)
    print(f"Created dummy image: {dummy_image_path}")

    extractor = ExtractColor()

    # Extract dominant colors
    dominant_colors_info, is_single_tone_verdict, processed_image_path = extractor.extractColor(dummy_image_path)

    print("\n--- Dominant Color Extraction Results ---")
    print("Dominant Colors Information:")
    for color in dominant_colors_info:
        print(f"  Hex: {color['hex']}, Percentage: {color['percentage']}%, RGB: {color['rbg_values']}")
    print(f"Is Single Tone: {is_single_tone_verdict}")
    print(f"Processed Image Path (might be resized): {processed_image_path}")

    # To get only the *most* dominant color, you can access the first element
    # of dominant_colors_info (assuming it's sorted by dominance by extcolors,
    # which it usually is).
    if dominant_colors_info:
        most_dominant_color = dominant_colors_info[0]
        print("\n--- Most Dominant Color ---")
        print(f"Hex: {most_dominant_color['hex']}")
        print(f"Percentage: {most_dominant_color['percentage']}%")
        print(f"RGB Values: {most_dominant_color['rbg_values']}")
    else:
        print("\nNo dominant colors found.")

    # Clean up the dummy image
    if os.path.exists(dummy_image_path):
        os.remove(dummy_image_path)
        print(f"\nCleaned up {dummy_image_path}")
```
